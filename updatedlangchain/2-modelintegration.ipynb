### Models Integration with OpenAI, Google Gemini and GROQ

```python
# 1 - LOAD API KEYS 
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
## 2 - init model
from langchain.chat_models import init_chat_model

# Option A: OpenAI (requires langchain-openai package)
# model = init_chat_model("gpt-4.1", model_provider="openai")

# Option B: Google Gemini (requires langchain-google-genai package)
# Note: Explicitly set model_provider="google_genai" to avoid defaulting to Vertex AI
model = init_chat_model("gemini-2.5-flash", model_provider="google_genai")

# Option C: GROQ (requires langchain-groq package)
# model = init_chat_model("llama-3.1-8b-instant", model_provider="groq")

model
```

```python
## 3 - invoke the model
response = model.invoke("hello, how are you?")
response
```

### GROQ Model Integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Which birds dont fly?")
response
```

```python
### 2nd way of calling models (usinh ChatGroq)

from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3-32b")
response = model.invoke("Why plant leaves became yellow?")
response
```

```python
### Streaming & Batch

# model.stream("Write me 200 words paragraph on Langchain")

for chunk in model.stream("Write me 200 words paragraph on Langchain"):
    print(chunk.text, end="|", flush=True)
```

```python
### Batch

responses = model.batch([
       "Why do airplanes fly?",
       "What is quantum computing?"
       "Which are useful crypto utility coins?"
    ],
    config={
        'max_concurrency' : 5 # Limit to 5 parallel calls
    }
)

for response in responses:
    print(response)
```
